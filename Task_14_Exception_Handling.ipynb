# Q.2)Write a Python program that defines a custom exception called NegativeValueError. The program should take an integer input from the user and raise a NegativeValueError if the input is negative.

```python
class NegativeValueError(Exception):
    pass
try:
    num=int(input("Enter the number:"))
    if num<=0:
        raise NegativeValueError
except  NegativeValueError:
    print("Negative value not permitted,please try again")
```

# Q.3)Write a Python program that takes user input to perform a mathematical operation. Handle ValueError if the input cannot be converted to a number and handle ZeroDivisionError if the user attempts to divide by zero.
* Write a Python program that takes a string as input and prints its length. Handle the TypeError exception if the user enters a non-string value.

```python
try:
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    print(a/b)
except ValueError:
    print("Error: Please enter numeric values only.")    
except ZeroDivisionError:
    print("Division by 0 is absurd and not permitted")
```

```python
try:
    text = input("Enter the string: ")
    if type(text) != str:
        raise TypeError("Input must be of string type!")

    print("Length of the string is:", len(text))

except TypeError as e:
    print("Error:", e)
```

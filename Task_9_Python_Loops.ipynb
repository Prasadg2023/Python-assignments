# Q.10) Count the number of vowels in a string using a for loop.
 * s2= "Hello World"


```python
s2 = "Hello World"
vowels = "aeiou"
count = 0
for ch in s2:
    if ch in vowels:
        count += 1
print("Number of vowels:", count)
```

# Q.12) Find all prime numbers between 1 and 50 using a for loop.


```python
for num in range(1,51):
    is_prime = True
    for i in range(2,num):
        if num%i==0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

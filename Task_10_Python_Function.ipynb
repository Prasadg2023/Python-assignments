# Q.1) Write a function to get the maximum of 3 numbers.

```python
a=10
b=20
c=2
def max_num(a,b,c):
    if a>= b and a>=c:
        return a
    elif b>= a and b>=c:
        return b
    else:
        return c
```

```python
print(max_num(a,b,c))
```

# Q.4) Construct your own function my_range() which acts as range() to print numbers in forward direction (step is positive) and backward direction(step is negative) with given start,stop and step as parameters.
* **eg:**
    * my_range(1,10,1), then o/p= 1,2,3,4,5,6,7,8,9 
    * my_range(1,10,2), then o/p= 1,3,5,7,9 
    * my_range(10,1,-1), then o/p = 10,9,8,7,6,5,4,3,2 
    * my_range(10,1,-2), then o/p = 10,8,6,4,2


```python
def my_range(start,stop,step):
    for i in range(start,stop,step):
        print(i,end=" ")
    
```

```python
my_range(1,10,1) #forward direction
```

```python
my_range(1,10,2) #forward direction
```

```python
my_range(10,1,-1) #backward direction
```

```python
my_range(10,1,-2) #backward direction
```

# Q.5) Multiply numbers with given start and stop (using loops)
* **Example:** 
    * start=1, stop= 5, then o/p: 120 
    * start=3, stop=5, then o/p: 60


```python
def multiplication_range(start,stop):
    mult=1
    i=start
    while i<= stop:
        mult*=i
        i+=1
    print(mult)
```

```python
multiplication_range(1,5)
```

```python
multiplication_range(3,5)
```

# Q.6) Construct a function for fibonacci series(fib()) using
* a.loops 
* b.recursion 
    * **Expected o/p :** 
        * fib(5), then o/p= 0 1 1 2 3 
        * fib(7), then o/p= 0 1 1 2 3 5 8


```python
#using loop
def fib(n):
    a,b=0,1
    fibseries=[]
    for i in range (n):
        fibseries.append(a)
        a,b=b,a+b
    print(fibseries)
```

```python
fib(5)
```

```python
fib(7)
```

```python
#using recurion
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1)+fibonacci(n-2)  #it print single number
#toprint series 
def fib(n):
    for i in range(n):
        print(fibonacci(i),end=" ")
```

```python
fib(5)
```

```python
fib(7)
```

# Q.7) Create a Factorial with loops.


```python
#using while loop
def factorial(n):
    fact = 1
    i = 1
    while i <= n:
        fact *= i
        i += 1
    print(fact)
    
```

```python
factorial(5)
```

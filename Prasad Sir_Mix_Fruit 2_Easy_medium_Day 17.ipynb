## Q.5) (Medium): Write a Python program to check if a number is a palindrome using a while loop.

```python
# to check if a number is a palindrome using a while loop.
num = 121
num2 = str(num)
while True:
    if num2 == num2[::-1]:
        print(f"{num} is palindrome")
    else:
        print(f"{num} is not palindrome")
    break
```

## Q.7) (Medium): Write a function that returns the Fibonacci sequence up to a given number of terms.

```python
# a function that returns the Fibonacci sequence up to a given number of terms.¶
def fibonacci(n):
    a = 0
    b = 1
    for _ in range(n):
        print(a, end = " ")
        a,b = b,a+b   
```

```python
fibonacci(5)
```

# 6. Lists
## Q.8) (Easy): Write a Python program to reverse a list without using the built-in reverse() function.

```python
# to reverse a list without using the built-in reverse() function.
a = [1,2,3,4,5]
rev=[]
for i in range(len(a),0,-1):
    rev.append(i)
print(rev)
```

# 9. File Handling
## Q.14) (Medium): Write a Python program to write a list of names to a file and then read and print those names.

```python
# to write a list of names to a file and then read and print those names.
names = ["Raj", "Patil", "Prasad", "Saurabh"]
with open("names.txt", "w") as f:
    for n in names:
        f.write(n+" ")

with open("names.txt", "r") as f:
    print(f.read())
```

# 10. List Comprehension & Lambda
## Q.15) (Medium): Using list comprehension and lambda, create a list of all numbers from 1 to 30 that are divisible by both 3 and 5.

```python
# a list of all numbers from 1 to 30 that are divisible by both 3 and 5.
a = list(filter(lambda x: x % 3 == 0 and x % 5 == 0, [i for i in range(1, 31)]))
print(a)
```

# 5. Count Words in a File
## Create a function that reads a file and returns the total number of words.

```python
# Count Words in a File¶
def count_words(filename):
    with open(filename, "r") as file:
        return len(file.read().split())

print(count_words("data.txt"))   
```

# Exception Handling

# 1. Write a program to divide two numbers input by the user. Use try-except to handle ZeroDivisionError.


```python
# handle ZeroDivisionError
try:
    a = int(input("Enter a number: "))
    b = int(input("Enter a number: "))
    result = a/b
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Result:", result)
```

# 7. Open a file for reading and ensure it gets closed using a finally block, even if an error occurs during reading.

```python
# handling error
filename = input("Enter a filename:")
try:
    file = open(filename,"r")
    content = file.read()
    print(content)
except Exception as e:
    print("An error occurred:", e)
finally:
     file.close()
     print("File closed successfully.")
```

# 10. Continuously ask the user to enter a number to add to a list. If they enter invalid input, catch the exception and continue the loop.

```python
# catch the exception and continue the loop
lst = []
while True:
    try:
        n = int(input("Enter a number: "))
        lst.append(n)
    except:
        print("Invalid input, try again.")
        continue
    print("List:", lst)
```

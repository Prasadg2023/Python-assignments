**Q.2)Set Operations Given two sets, write a Python program to find the union, intersection, and difference of these sets."**


```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

union_set = set1 | set2
intersection_set = set1 & set2
difference_set = set1 - set2

print(union_set)
print(intersection_set)
print(difference_set)
```

**Q.4)Set Operations with Lists Create a Python list with duplicate elements and convert it into a set to remove duplicates. Then, write a program to count how many duplicates were removed.**


```python
numbers = [1, 2, 3, 2, 4, 3, 5, 1]
unique_numbers = set(numbers) 
duplicates_removed = len(numbers) - len(unique_numbers) # how many duplicates counting
print(numbers)
print(unique_numbers)
print(duplicates_removed)


```

**Q.5)Explain the concept of Sets and their properties in Python. Provide a one-line definition of each set method with examples.**


A Set in Python is an unordered collection of unique elements.
It is used to store multiple items in a single variable, where duplicates are automatically removed.

**Properties:**

* **Unordered** – Elements have no defined order.

* **Unindexed** – You can’t access elements using index numbers.

* **Mutable** – You can add or remove items after creation.

* **No Duplicates** – Each element appears only once.

**Method and their definitio with example:**


* **add()**– Adds a single element to the set.
    * Example: s.add(4)

* **update()** – Adds multiple elements (from list, set, etc.) to the set.
    * Example: s.update([5, 6])

* **remove()** – Removes the specified element; raises an error if not found.
    * Example: s.remove(3)

* **discard()** – Removes the specified element; does not raise an error if missing.
    * Example: s.discard(10)

* **pop()** – Removes and returns a random element from the set.
    * Example: s.pop()

* **clear()** – Removes all elements from the set.
    * Example: s.clear()

* **copy()** – Returns a shallow copy of the set.
    * Example: new_set = s.copy()

* **union()** – Returns a new set with all elements from both sets.
    * Example: s1.union(s2)

* **intersection()** – Returns a set with common elements of both sets.
    * Example: s1.intersection(s2)

* **difference()** – Returns elements that exist in one set but not in another.
    * Example: s1.difference(s2)

* **symmetric_difference()**– Returns elements present in either set, but not both.
    * Example: s1.symmetric_difference(s2)

* **issubset()** – Returns True if all elements of one set are in another.
    * Example: s1.issubset(s2)

* **issuperset()** – Returns True if one set contains all elements of another.
    * Example: s1.issuperset(s2)

* **isdisjoint()** – Returns True if two sets have no common elements.
    * Example: s1.isdisjoint(s2)

# Q.1) Write a function called "count_vowels" that takes a string as input and returns the number of vowels (a, e, i, o, u) in that string.


```python
def count_vowels(string):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in string:
        if ch in vowels:
            count += 1
    return count
text=input("Enter a string: ")
print("Number of vowels:",count_vowels(text))
```

# Q.2) Write a Python function to sum all the numbers in a list.(Without using built-in function sum).

```python
def sum_of_numbers(numbers):
   
    total=0
    for num in numbers:
        total += num
    return total

nums = [1,2,3,4,5]
print("sum of numbers:",sum_of_numbers(nums))
```

# Q.3) Write a Python function to check whether a number falls within a given range. The function should take three arguments: start, stop, and number. The function should return True if the number is within the range [start, stop] (inclusive), and False otherwise.


```python
def in_range(start,stop,number):
    if number in range(start,stop):
        return True
    else:
        return False
```

```python
in_range(1,10,15)
```

```python
in_range(1,10,5)
```

# Q.4) Write a Python function that accepts a string and counts the number of upper and lower case letters


```python
def count_letters(string):
    upper=0
    lower=0
    for ch in string:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1
    print("number of uppercase letters in string:",upper)
    print("number of lowercase letters in string:",lower)
count_letters("Innomatics is the best for Data Science")
```

# Q.5) Write a function is_prime(n) that returns True if the number n is prime, otherwise False.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2,int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
```

```python
is_prime(8)
```

# Q.6) Write a function is_anagram(s1, s2) that takes two strings s1 and s2 and returns True if they are anagrams of each other, otherwise False.
* is_anagram("listen", "silent")  # Output: True
* is_anagram("hello", "world")    # Output: False


```python
def is_anagram(s1,s2):
    return sorted(s1)==sorted(s2)
```

```python
is_anagram("listen", "silent")
```

```python
is_anagram("hello", "world")   
```

# Q.7) Write a function sum_of_digits(n) that takes an integer n and returns the sum of its digits.
* sum_of_digits(1234)  # Output: 10


```python
def sum_of_digit(n):
    return sum(int(digit) for digit in str(n))
```

```python
sum_of_digit(1234)
```

```python
sum_of_digit(1234989)
```

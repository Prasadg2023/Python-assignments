**Q.1)Define Strings and its properties.**



**Definition:**


        In Python, a string is a sequence of characters used to represent text. String is immutable in nature that once created it cannot be changed.It must be written in Quotes.For example a="Hello".

**Properties:**

**1.Immutability:** Strings in Python are immutable, that means their content cannot be changed after creation. Any operation that occrus to modify a string e.g., replace(), upper(), concatenation actually creates a new string with the certain changes that desired changes.

**2.Sequence of Characters:** Strings are ordered sequences of characters. This allows for indexing and slicing to access individual characters.  

**3.Defined by Quotes:** Strings are created by enclosing sequences of characters within single quotes ('...'), double quotes ("..."), or triple quotes ('''...''' or """..."""). Triple quotes are specifically useful for multi-line strings.

**4.No Separate Character Type:** In Python, there is no any "character" data type. A single character can be written as string. 

**5.Operable with Operators:** Various type of operations takes place.Such as,it concatenated using + and reapeated using * operator.

**Q.10)Explore all string methods from documentation or use dir() function and write one line explanation for each method.**


**The dir() function in Python can be used to list the attributes and methods of an object. When used with a string, it will return a list of all available string methods. Here are some of the most common string methods with a one-line explanation for each:**
* **capitalize():** Returns a copy of the string with its first character capitalized and the rest lowercase.
* **center(width[, fillchar]):** Returns a centered string of a specified width, padded with the specified fillchar (default is a space).
* **count(sub[, start[, end]]):** Returns the number of non-overlapping occurrences of sub in the string.
* **endswith(suffix[, start[, end]]):** Returns True if the string ends with the specified suffix, otherwise False.
* **find(sub[, start[, end]]):** Returns the lowest index in the string where sub is found, or -1 if not found.
* **isalnum():** Returns True if all characters in the string are alphanumeric and there is at least one character, otherwise False.
* **isalpha():** Returns True if all characters in the string are alphabetic and there is at least one character, otherwise False.
* **isdigit():** Returns True if all characters in the string are digits and there is at least one character, otherwise False.
* **nislower():** Returns True if all cased characters in the string are lowercase and there is at least one cased character, otherwise False.
* **isspace():** Returns True if there are only whitespace characters in the string and there is at least one character, otherwise False.
* **isupper():** Returns True if all cased characters in the string are uppercase and there is at least one cased character, otherwise False.
* **join(iterable):** Returns a string which is the concatenation of the strings in iterable.
* **lower():** Returns a copy of the string with all cased characters converted to lowercase.
* **replace(old, new[, count]):** Returns a copy of the string with all occurrences of old replaced by new.
* **startswith(prefix[, start[, end]]):** Returns True if the string starts with the specified prefix, otherwise False.
* **upper():** Returns a copy of the string with all characters converted to uppercase.

**Q.1)Explain about idenfiers and their rules**

        An identifier is a name to identify a variable,function,class,module and other objects
    in python.It helps to uniquely identify these entities with program which makes code easily 
    readable and managable

**Rules for creating Python identifiers:**

* Identifiers can be a combination of letters a-z or A-Z, digits 0-9, and the underscore _. For exapmle-a=10,a1=13
* An identifier cannot start with a digit. for example- 1a=10
* In python 35 keywords are present and these keywords cannot be usd as identifiers
* Identifiers are case-sensitive,that you should write a code carefully
* Special characters like !, @, #, $, % are not allowed in identifiers.

**Q.3)Explain why Python treats them as invalid identifiers.Explain?**

      Python treats the invalid variables from the last question as such because they spoils the fundamental rules for naming 
      identifiers in python as we discussed earlier.some reasons are as follows:

* **Starting with a digit:** 123abc is invalid because an identifier cannot begin with a number.
* **Containing spaces:** my variable is invalid because identifiers must be single or continuous words.
* **Containing special characters:** "@count=""hi", "if condition?""Innomatics", and "class=""Hyderabad***" are invalid because they occurs special
  characters like @, ?, * that are not allowed in identifiers.
* **Using Python's keywords:** "for = ""True", "if condition?""Innomatics", and "class=""Hyderabad***" are invalid because for, if, and class are
  keywords  in Python and cannot be used as variable names.
* **Written in quotes:** All the invalid variables that are written in quotes are invalid because identifiers are not string literals.

**Q.1)Define Dictionary, its properties and examples for each method.**


A Dictionary in Python is an unordered collection of key–value pairs and where
each value is accessed using its unique key instead of an index like other collection datatypes 

**Properties:**

* **Key–Value Pair:** Each item has a unique key and a corresponding value.

* **Unordered:** The order of items is not guaranteed.

* **Mutable:** You can change, add, or remove items after creation.

* **Keys must be unique:** No two keys can be the same.

* **Keys must be immutable:** Keys can be strings, numbers, or tuples, but not lists or sets.

**Methods:**
* **1. clear()** – Removes all items from the dictionary.
    * Example: student.clear()
* **2. copy()** – Returns a shallow copy of the dictionary.
    * Example:new_dict = student.copy()


* **3. fromkeys()** – Creates a new dictionary from given keys with a specified value.
    * Example:keys = ['a', 'b', 'c']
        new_dict = dict.fromkeys(keys, 0)


* **4. get()** – Returns the value for a key, or None if key not found.
    * Example:student.get("age")


* **5. items()** – Returns all key–value pairs as a view object.
    * Example:student.items()


* **6. keys()** – Returns all keys in the dictionary.
    * Example:student.keys()


* **7. values()** – Returns all values in the dictionary.
    * Example:student.values()


* **8. pop()**– Removes and returns the value of the specified key.
    * Example:student.pop("age")


* **9. popitem()** – Removes and returns the last inserted key–value pair.
    * Example:student.popitem()


* **10. setdefault()** – Returns the value of a key; inserts key with a value if not found.
   * Example:student.setdefault("city", "Pune")


* **11. update()**– Updates the dictionary with items from another dictionary or key–value pairs.
    * Example:student.update({"age": 21})


**Q.2)Explore all the methods in the dictionary and write example for each method**


* **1. clear():**
 * Removes all items from the dictionary.

```python
#Example
student = {"name": "Prasad", "age": 22}
student.clear()
print(student)      
```

**2. copy():**
* Returns a shallow copy of the dictionary.

```python
#Example
student = {"name": "Prasad", "age": 22}
new_dict = student.copy()
print(new_dict)
```

**3. fromkeys():**
* Creates a new dictionary from given keys with a common value.

```python
keys = ['a', 'b', 'c']
new_dict = dict.fromkeys(keys, 0)
print(new_dict)
```

**4. get()**
* Returns the value for a given key; returns None if key not found.

```python
student = {"name": "Ravi", "age": 20}
print(student.get("age"))     
print(student.get("city"))
```

**5. items()**
* Returns all key–value pairs as a view object.

```python
student = {"name": "Prasad", "age": 22}
print(student.items())    
```

 **6. keys():**
* Returns all keys in the dictionary.

```python
student = {"name": "Prasad", "age": 22}
print(student.keys())     
```

**7. values():**
* Returns all values in the dictionary.

```python
student = {"name": "Ravi", "age": 20}
print(student.values())   
```

**8. pop():**
* Removes and returns the value for the specified key.

```python
student = {"name": "Prasad", "age": 22}
removed = student.pop("age")
print(removed)    
print(student)    
```

**9. popitem():**
* Removes and returns the last inserted key–value pair.

```python
student = {"name": "Prasad", "age": 22, "city": "Pune"}
item = student.popitem()
print(item)       
print(student)    
```

**10. setdefault():**
* Returns the value of a key; inserts the key with a default value if not found.

```python
student = {"name": "Prasad", "age": 22}
student.setdefault("city", "Mumbai")
print(student)   
```

**11. update():**
* Updates the dictionary with items from another dictionary or key–value pairs.

```python
student = {"name": "Prasad", "age": 22}
student.update({"age": 21, "city": "Pune"})
print(student)   
```

**12. len() (Built-in function)**
* Returns the number of key–value pairs in the dictionary.

```python
student = {"name": "Ravi", "age": 20, "city": "Pune"}
print(len(student))   
```

**13. del (It is Statement, not a method):**
* Deletes a key–value pair or the entire dictionary.

```python
student = {"name": "Ravi", "age": 20}
del student["age"]
print(student)        
```

**Q.3)Create a Python program that defines a dictionary to store information about your favorite book. Include the title, author, publication year, and genre as key-value pairs. Print out the dictionary.and do at least 5 dictionary methods on it.**


```python
# Define a dictionary for favorite book
book = {
    "title": "To the last bullet",
    "author": "Vinitha Deshmukh",
    "publication_year": 2009,
    "genre": "Inspiring"
}
# Print the dictionary
print("Original Dictionary:")
print(book)
# get() - # value of a specific key
print("\n1. Title of the book:", book.get("title"))
# keys() # Display all keys
print("\n2. All Keys:", book.keys())
# values() #Display all values
print("\n3. All Values:", book.values())
# update() #Add a new key-value pair
book.update({"language": "Portuguese"})
print("\n4. After Adding Language:", book)
# pop() #Remove a key-value pair
book.pop("publication_year")
print("\n5. After Removing Publication Year:", book)
```

**Q.4) Write a Python program that simulates a simple dictionary for translating English words to Spanish. Define a dictionary containing a few English words as keys and their corresponding Spanish translations as values. Allow users to enter an English word, and then display its Spanish translation if it exists in the dictionary.(use google for English to Spanish translator)**


```python
# English to Spanish Dictionary
translator = {
    "hello": "hola",
    "goodbye": "adiós",
    "please": "por favor",
    "thank you": "gracias",
    "sorry": "lo siento",
    "friend": "amigo",
    "love": "amor",
    "book": "libro",
    "food": "comida",
    "water": "agua"
}
# english word from user
word = input("Enter an English word to translate: ").lower()
# if any translation takesplace it shows
if word in translator:
    print(f"The Spanish translation of '{word}' is '{translator[word]}'.")
else:
    print(f"Sorry, the word '{word}' is not in the dictionary.")
```

**Q.5) Create a Python program that stores the grades of five students in a dictionary. Allow users to input the names of the students and their respective grades. Calculate and display the average grade of the students.**


```python
grades = {}

for i in range(5):
    name = input("Enter student name: ")
    grade = float(input("Enter grade: "))
    grades[name] = grade

average = sum(grades.values()) / 5
print("Average grade:", average)
```

**Q.6)Build a basic contact list using a dictionary. Initially, the dictionary can be empty. Allow users to add new contacts with names and phone numbers. Users should also be able to search for contacts by name and display their phone numbers(use in operator).**


```python
contacts = {}

name = input("Enter name: ")
phone = input("Enter phone number: ")
contacts[name] = phone

search = input("Enter name to search: ")

if search in contacts:
    print("Phone number:", contacts[search])
else:
    print("Contact not found.")
```

**Q.7)Write a Python program that defines a dictionary containing the names of fruits as keys and their corresponding colors as values (e.g., 'apple': 'red').Do atleast 5 dictionary methods on it.**

```python
# Define dictionary
fruits = {
    'apple': 'red',
    'banana': 'yellow',
    'grape': 'purple',
    'orange': 'orange',
    'kiwi': 'green'
}
# keys() - get all fruit names
print("Fruit names:", fruits.keys())
# values() - get all colors
print("Fruit colors:", fruits.values())
# get() - get color of a specific fruit
print("Color of apple:", fruits.get('apple'))
# update() - add a new fruit
fruits.update({'mango': 'yellow'})
print("After adding mango:", fruits)
# pop() - remove a fruit
fruits.pop('grape')
print("After removing grape:", fruits)
```

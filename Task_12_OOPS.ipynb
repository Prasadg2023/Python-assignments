* Example: 
class Person:

    #code
    
class Address:

    #code
    
class Employee(Person, Address):

    #code
* Example Output - Employee: Sirisha,  Age: 25,  City: Delhi,  Country: India,  Job: Data Scientist


```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

class Address:
    def __init__(self,city,country):
        self.city=city
        self.country=country

class Employee(Person,Address):
    def __init__(self,name,age,city,country,job):
        Person.__init__(self,name,age)
        Address.__init__(self,city,country)
        
        self.job=job

    def display_employee_info(self):
        print(f"Employee:{self.name}, Age:{self.age}, City:{self.city}, Country:{self.country}, Job:{self.job}")
        
    
```

```python
employee1=Employee("Sirisha",25,"Delhi","India","Data Scientist")
```

```python
employee1.display_employee_info()
```

# Q.10) Create an abstract class Shape with an abstract method calculate_area(). Define two subclasses, Rectangle and Circle, that inherit from Shape and implement the calculate_area() method.


```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def calculate_area(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return math.pi * (self.radius ** 2)

    
```

```python
shape=Rectangle(10,12)
```

```python
shape.calculate_area()
```

```python
shape1=Circle(24)
```

```python
shape1.calculate_area()
```

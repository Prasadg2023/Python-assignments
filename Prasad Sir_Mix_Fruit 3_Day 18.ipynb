## Q.1) (Medium): Write a function that takes a list of strings and returns a new list with all strings converted to uppercase and sorted alphabetically.


```python
# function that takes a list of strings and returns a new list with all strings converted to uppercase and sorted alphabetically.
def str_lst(lst):
    upper_lst = []
    for w in lst:
        upper_lst.append(w.upper())
    return sorted(upper_lst)
```

```python
str_lst(["Prasad","Raja","Yash"])
```

# 2. Control Flow
## Q.2) (Medium): Write a Python program that checks whether a number is a prime or not (input from user).

```python
# checks whether a number is a prime or not (input from user).
a = int(input("Enter a number: "))
if a <= 1:
    print("Not a prime number")
else:
    for i in range(2, a):
        if a % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```

# 7. Strings
## Q.7) (Medium): Write a function that takes a string and returns a new string with all vowels removed.

```python
# a string and returns a new string with all vowels removed.
def string(s):
    n_string = ""
    vowels = "aeiouAEIOU"
    for ch in s:
        if ch not in vowels:
            n_string += ch
    print(n_string)    
```

# 10. OOP-Class and Object
## Q.10) (Medium): Create a class Rectangle with attributes length and width. Add methods to calculate area and perimeter.

```python
# calculate area and perimeter.
class Rectangle:
    def __init__(self,l,w):
        self.l = l
        self.w = w
    def cal_area_peri(self):
        area = self.l * self.w
        perimeter = 2 * (self.l + self.w)
        print(f"Area of Rectangle is: {area} and Perimetere of the Rectangle is: {perimeter}")
```

```python
a = Rectangle(2,4)
```

```python
a.cal_area_peri()
```

# 11. OOP-Inheritance
## Q.11) (Medium): Create a class Person with name and age, and a subclass Student that adds a student ID. Print their details using a method.

```python
# Print details using a method
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self,name,age,student_ID):
        super().__init__(name,age)
        self.student_ID = student_ID
    
    def student_details(self):
        print(f"{self.name} is of age {self.age} has student ID {self.student_ID}")
```

```python
a = Student("Prasad",23,2130421995037)
```

```python
a.student_details()
```

# 13. Nested Lists
## Q.13) (Medium): Write a function that takes a nested list of integers and returns the sum of all integers.

```python
# sum of all integers.¶
def Nested_lst(lst):
    total = 0
    for item in lst:
        if type(item) == list:
            total += sum(item)
        else:
            total += item
    print(total)   
```

```python
Nested_lst([1, [2, 3], [4]])
```

## Q.14) (Medium): Write a Python function that takes a list of tuples (name, age) and returns the list sorted by age.

```python
#sorted by age
def sort_by_age(pairs):
    print(sorted(pairs, key=lambda x: x[1]))
```

# 15. Datetime
## Q.15) (Medium): Write a Python function that takes two date strings in the format YYYY-MM-DD and prints how many days apart they are.

```python
# how many days apart they are
from datetime import datetime

def days_apart(d1, d2):
    date1 = datetime.strptime(d1, "%Y-%m-%d")
    date2 = datetime.strptime(d2, "%Y-%m-%d")
    diff = abs((date1 - date2).days)
    print(diff, "days apart")
```

```python
days_apart("2025-01-10", "2025-01-25")
```
